```python
def give_me_a_key():
    key=input("Enter any key on the keyboard:")
    result=''
    try:
      num=int(key)
    except:
      if key.isalpha():
        result=key.upper()
      else:
        result=key
    else:
      result=num**2

    finally:
      return result

print(give_me_a_key())


```

```python
def calculate(*args,**kwargs):
  operations=kwargs.get("operation","add")#if the user doesnot provide any operation,By default add will be the operation
  if not args:  # handle empty input
        print("No numbers provided!")
        return
  if operations=="add":
    total= sum(args)
  elif operations=="multiply":
    total=1
    for i in args:
      total*=i

  else:
    print("Invalid operation")
  print(f"{operations}:{total}")
calculate(1,2,3,4,operation="multiply")
calculate(1,2,5)
```

```python
if __name__ == '__main__':
    n = int(input())
    student_marks = {}
    for _ in range(n):
        name, *line = input().split()
        scores = list(map(float, line))
        student_marks[name] = scores
    query_name = input()
    marks=student_marks[query_name]
    for i in marks:
      avg=sum(marks)/len(marks)
    print(f"{avg}:.2f")
```

```python
import time
class PinException(Exception):
  pass

class WithdrawException(Exception):
  pass
attempts=1
def withdraw():
  global attempts
  correct_pin=0000
  balance=10000
  pin=int(input("Enter your pin:"))
  if pin==correct_pin:
    try:
      amt=float(input("Enter the amount to withdraw:"))
      bal=balance-amt
      if bal<1000:
        raise WithdrawException("Insufficient balance")
      print(bal)
    except WithdrawException as e:
        print(e)
  else:
    if (input(("Incorrect pin,Do you want to try again:y/n=")).lower())=="y":
      attempts +=1
      print(f"Attempts left:{3-attempts}")
      try:
        if attempts==3:
          raise PinException("Too many attempts,Your account is blocked for an hour")
      except PinException as p:
            print(p)
            time.sleep(3600)
      else:
          withdraw()
    else:
      print("Thank You")
      return

withdraw()

```
